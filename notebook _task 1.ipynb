## 2. Python Basics for Data Science
### 2.1 Variables and data types

```python
name = "Keerthana"        # str
age = 20                  # int (sample value, change to yours)
cgpa = 9.13               # float
is_student = True         # bool

for v in (name, age, cgpa, is_student):
    print(f"{v!r:12} -> {type(v).__name__}")

# Arithmetic and type conversion
a, b = 17, 5
print("a + b =", a + b, "| a / b =", a / b, "| a // b =", a // b, "| a % b =", a % b, "| a ** b =", a ** b)
print(int("42") + 8, float("3.5") * 2, str(100) + " marks")
```

### 2.2 Conditional statements and loops

```python
def grade(marks):
    if marks >= 90: return "A+"
    elif marks >= 80: return "A"
    elif marks >= 70: return "B"
    elif marks >= 60: return "C"
    return "Fail"

for m in [95, 83, 71, 64, 40]:
    print(m, "->", grade(m))

# for loop with range
print("Squares:", [i**2 for i in range(1, 6)])

# while loop
n, total = 1, 0
while n <= 5:
    total += n
    n += 1
print("Sum of 1..5 =", total)

# break / continue
for i in range(1, 10):
    if i % 2 == 0: continue
    if i > 7: break
    print("odd:", i, end="  ")
```

### 2.3 Functions

```python
def mean(values):
    """Return the arithmetic mean of a list of numbers."""
    return sum(values) / len(values)

def describe(values):
    """Return min, max, mean and range as a dictionary."""
    return {"min": min(values), "max": max(values),
            "mean": round(mean(values), 2), "range": max(values) - min(values)}

scores = [78, 85, 92, 66, 88, 74]
print(describe(scores))

# default and keyword arguments, lambda
def power(x, p=2): return x ** p
print(power(4), power(2, p=10))
square = lambda x: x * x
print(list(map(square, [1, 2, 3])))
```

### 2.4 Lists

```python
subjects = ["Maths", "Python", "Statistics"]
subjects.append("Machine Learning")
subjects.insert(1, "DBMS")
subjects.remove("Maths")
print(subjects)
print("First:", subjects[0], "| Last:", subjects[-1], "| Slice:", subjects[1:3])

marks = [88, 92, 79, 95, 67]
marks.sort(reverse=True)
print("Sorted:", marks, "| Top 3:", marks[:3])

# List comprehension with a condition
passed = [m for m in marks if m >= 70]
print("Passed:", passed)
```

### 2.5 Dictionaries

```python
student = {"name": "Keerthana R", "dept": "AI & DS", "cgpa": 9.13}
student["email_verified"] = True    # add
student["cgpa"] = 9.13               # update
print(student)
print("Keys:", list(student.keys()))
print("Dept:", student.get("dept"), "| Missing:", student.get("phone", "N/A"))

# Dictionary of lists: word frequency counter
text = "data science uses data to find patterns in data"
freq = {}
for word in text.split():
    freq[word] = freq.get(word, 0) + 1
print(freq)
print("Most common:", max(freq, key=freq.get))
```

### 2.6 File handling

```python
# Write a text file
with open("notes.txt", "w") as f:
    f.write("Python is great for data science.\n")
    f.write("Colab and Jupyter make experimenting easy.\n")

# Append
with open("notes.txt", "a") as f:
    f.write("File handling uses open(), read(), write().\n")

# Read whole file and line by line
with open("notes.txt") as f:
    for i, line in enumerate(f, 1):
        print(i, line.strip())

# Work with CSV
import csv
rows = [["name", "marks"], ["Asha", 88], ["Ravi", 74], ["Meena", 93]]
with open("marks.csv", "w", newline="") as f:
    csv.writer(f).writerows(rows)

with open("marks.csv") as f:
    data = list(csv.DictReader(f))
print(data)
print("Average:", sum(int(r["marks"]) for r in data) / len(data))

# Handle errors safely
try:
    open("does_not_exist.txt")
except FileNotFoundError as e:
    print("Caught:", e)
```

## 4. Data Science Concepts

**Data Science** combines programming, statistics and domain knowledge to extract insight from data.

**Typical workflow**
1. **Define the problem** → what question are we answering?
2. **Collect data** → files, databases, APIs, web scraping
3. **Clean data** → handle missing values, duplicates, wrong types
4. **Explore (EDA)** → summary statistics and visualisations
5. **Model** → machine learning / statistical models
6. **Evaluate & communicate** → metrics, charts, decisions

**Core libraries:** NumPy (arrays), Pandas (tables), Matplotlib/Seaborn (plots), Scikit-learn (ML).

### 4.1 Mini data analysis with Pandas

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "student": ["Asha", "Ravi", "Meena", "Karthik", "Divya", "Suresh"],
    "python":  [88, 74, 93, 67, 81, None],
    "stats":   [79, 65, 90, 72, 85, 70],
    "dept":    ["AIDS", "CSE", "AIDS", "CSE", "AIDS", "CSE"],
})
print(df, "\n")
print("Missing values:\n", df.isna().sum(), "\n")

# Cleaning: fill missing with column mean
df["python"] = df["python"].fillna(df["python"].mean())
df["average"] = df[["python", "stats"]].mean(axis=1).round(1)

print(df.describe().round(2), "\n")
print(df.groupby("dept")["average"].mean().round(2))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
df.plot.bar(x="student", y=["python", "stats"], ax=ax[0], title="Marks by student")
df.groupby("dept")["average"].mean().plot.bar(ax=ax[1], color=["#4c78a8", "#f58518"], title="Average by department")
plt.tight_layout()
plt.show()
```
